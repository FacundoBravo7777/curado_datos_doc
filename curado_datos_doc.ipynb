```python
!pip install biopython # Install biopython using pip
# eliminar seqs duplicadas
from Bio import SeqIO
from google.colab import drive

# Monta Google Drive (si aún no lo has hecho)
drive.mount('/content/drive')

# Rutas de archivos
ruta_archivo_fasta_entrada = '/content/drive/MyDrive/doctorado/dataset_doc/b10/uniprotKB/uniprot_b10_todas_primerfiltro.fasta'
ruta_archivo_fasta_salida = '/content/drive/MyDrive/doctorado/dataset_doc/b10/uniprotKB/uniprot_b10_todas_primerfiltro_sindup.fasta'

try:
    # Diccionario para almacenar las secuencias únicas
    secuencias_unicas = {}

    # Leer el archivo FASTA de entrada y almacenar las secuencias únicas
    for record in SeqIO.parse(ruta_archivo_fasta_entrada, "fasta"): # Usa SeqIO.parse
        if str(record.seq) not in secuencias_unicas:
            secuencias_unicas[str(record.seq)] = record

    # Escribir las secuencias únicas en el archivo FASTA de salida
    with open(ruta_archivo_fasta_salida, 'w') as archivo_salida:
        for record in secuencias_unicas.values():
            SeqIO.write(record, archivo_salida, "fasta")

    print(f"Archivo FASTA sin duplicados creado exitosamente en: {ruta_archivo_fasta_salida}")

except FileNotFoundError:
    print(f"Error: Archivo no encontrado en la ruta: {ruta_archivo_fasta_entrada}")
except Exception as e:
    print(f"Error inesperado: {e}")
```

```python
# primer filtro
import re
from google.colab import drive

# Monta Google Drive (si aún no lo has hecho)
drive.mount('/content/drive')

# Rutas de archivos
ruta_archivo_fasta_entrada = '/content/drive/MyDrive/doctorado/dataset_doc/b10/uniprotKB/uniprot_b10_todas.fasta'
ruta_archivo_fasta_salida = '/content/drive/MyDrive/doctorado/dataset_doc/b10/uniprotKB/uniprot_b10_todas_primerfiltro.fasta'

try:
    with open(ruta_archivo_fasta_entrada, 'r') as archivo_entrada, open(ruta_archivo_fasta_salida, 'w') as archivo_salida:
        escribir_secuencia = False  # Inicializa la variable para controlar la escritura de la secuencia
        for linea in archivo_entrada:
            if linea.startswith('>'):  # Línea de descripción
                # Filtro por palabras clave en la descripción (case-insensitive)
                if not re.search(r'(partial|hypothetical|predicted|uncharacterized|low quality|unnamed)', linea, re.IGNORECASE):
                    archivo_salida.write(linea)
                    escribir_secuencia = True
                else:
                    escribir_secuencia = False
            elif linea.strip() and escribir_secuencia:  # Línea de secuencia (no vacía)
                # Filtro por "X" o "N" en la secuencia (case-sensitive)
                if not re.search(r'[X]', linea):  # Si no encuentra X o N
                    archivo_salida.write(linea)
                else:
                  escribir_secuencia = False # Si encuentra X o N, no se escribe la secuencia

    print(f"Archivo FASTA filtrado creado exitosamente en: {ruta_archivo_fasta_salida}")

except FileNotFoundError:
    print(f"Error: Archivo no encontrado en la ruta: {ruta_archivo_fasta_entrada}")
except Exception as e:
    print(f"Error inesperado: {e}")
```
